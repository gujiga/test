```python
def uses_none(word, forbidden):
    for i in word:
        flag=0
        for j in forbidden:
            if(i==j):
                flag=1;
                break
        if(flag==0):
            return False
    return True
print(uses_none("banana","ban"))
print(uses_none("apple","apl"))
```

```python
def uses_all(word, required):
    for i in required:
        flag=0
        for j in word:
            if(i==j):
                flag=1;
                break
        if(flag==0):
            return False
    return True
print(uses_all("banana","ban"))
print(uses_all("apple","api"))
    
```

```python
def check_word(word, available, required):
    if(len(word)<4):
        return False

    available=available.lower();
    required=required.lower();
    for i in word:
        flag=0
        for j in available:
            if(i==j):
                flag=1;
                break;
        if(flag==0):
            return False;
    flag=0
    for i in word:
        if(i==required):
            flag=1
    if(flag==0):
        return False
    return True
print(check_word('color', 'ACDLORT', 'R'))
print(check_word('ratatat', 'ACDLORT', 'R'))
print(check_word('rat', 'ACDLORT', 'R'))
print(check_word('told', 'ACDLORT', 'R'))
print(check_word('bee', 'ACDLORT', 'R'))
```
