```python
def A(m,n):
    if m==0:
        return n+1
    elif m>0 and n==0:
        return A(m-1,1)
    else:
        return A(m-1,A(m,n-1))
A(3,3)
```

```python
def gcd(a,b):
    if b==0:
        return a
    else:
        return gcd(b,a%b)
print(gcd(30,45))
print(gcd(4,6))
```

```python
def head(input_filename, num_lines, output_filename):
    input_file = open(input_filename, 'r')
    lines = []
    count = 0
    while count < num_lines:
        line = input_file.readline()
        lines.append(line)
        count += 1
    input_file.close()
    output_file = open(output_filename, 'w')
    output_file.writelines(lines)
    output_file.close()
head("input.txt",5,"output.txt")
```

```python
def is_palindrome(s):
    return list(reversed(s)) == list(s)
print(is_palindrome("aba"))
print(is_palindrome("abc"))
```

```python
def reverse_sentence(s):
    ls=s.split()
    ls = ls[::-1]
    ans = ""
    for i in ls:
        ans = ans + i + " "
    return ans
print(reverse_sentence("Reverse this sentence"))
```

```python
def has_duplicates(s):
    ls = list(s)
    ls.sort()
    siz = len(ls)
    for i in range(siz-1):
        if ls[i]==ls[i+1]:
            return True
    return False
print(has_duplicates("abc"))
print(has_duplicates("abca"))
```

```python
def find_repeats(mp):
    ans = []
    for x,y in mp.items():
        if y>1:
            ans.append(x)
    return ans
counts = {'a': 3, 'b': 1, 'c': 2, 'd': 4}
print(find_repeats(counts))
```

```python
def add_counters(c1, c2):
    ans = c1
    for letter, count in c2.items():
        if letter in ans:
            ans[letter] += count
        else:
            ans[letter] = count
    return ans
c1 = {'a': 3, 'b': 1, 'c': 2, 'd': 4}
c2 = {'a': 1, 'b': 2, 'c': 3, 'd': 4, 'e': 5}
print(add_counters(c1,c2))
```
